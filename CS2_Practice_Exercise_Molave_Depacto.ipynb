##**1. Personal Savings Tracker**

###**Scenario**
A student tracks money deposited into a savings fund. The program accepts each deposit until the student enters 0, then calculates the total savings and checks whether the target has been reached.


###**Requirements**
- Create a function update_savings(current_total, deposit) that returns the updated savings total.
- Create a function check_goal(total, target) that returns "Goal reached!" if the total is greater than or equal to the target, or "Keep saving!" otherwise.
- Ask the user to enter a savings target.
- Use a while loop to accept deposits until the user enters 0.
- Use the return value of update_savings() to update the running total after each deposit.
- After the loop, display the total savings, the remaining amount needed if the goal has not been reached, and the returned goal status.

###**Sample Output**
#####Enter savings target: ₱1000
#####**Enter deposit (0 to finish):** ₱200
#####Updated savings: ₱200.00
#####**Enter deposit (0 to finish):** ₱350
#####Updated savings: ₱550.00
#####**Enter deposit (0 to finish):** ₱500
#####Updated savings: ₱1050.00
#####**Enter deposit (0 to finish):** ₱0
#####Total savings: ₱1050.00
#####**Goal reached!**

> **Concepts:** Passing updated values between function calls, numeric and string return values, while loops, sentinel-controlled iteration, and accumulators.

```python
def update_savings(total, deposit):
    return total + deposit

def check_goal(total, target):
    if total >= target:
        return "Goal reached!"
    return "Keep saving!"

target = float(input("Enter savings target: ₱"))
total = 0

while True:
    deposit = float(input("Enter deposit (0 to finish): ₱"))

    if deposit == 0:
        break

    total = update_savings(total, deposit)
    print("Updated savings: ₱", total)

print("Total savings: ₱", total)

if total < target:
    print("Remaining amount: ₱", target - total)

print(check_goal(total, target))
```

##**2. Classroom Energy Usage Analyzer**
###**Scenario**
A class monitors the estimated energy consumption of electrical devices used during a school day. The program calculates the energy consumed by each device and summarizes the readings without storing them in an array or list.
###**Requirements**
- Create a function calculate_energy(power_watts, hours_used) that returns energy consumption in kilowatt-hours (kWh).
- Create a function classify_usage(energy) that returns:
- "Low" if energy consumption is below 0.5 kWh.
- "Moderate" if energy consumption is from 0.5 to 2.0 kWh, inclusive.
- "High" if energy consumption is above 2.0 kWh.
- Ask the user how many devices will be recorded.
- Use a for loop to input each device's power rating in watts and hours of use.
- Calculate and display each device's energy consumption and classification.
- Track the total energy consumption using individual variables.
- Display the total energy consumption and the name of the device with the highest consumption.
###**Sample Output**
######How many devices? 3
######**Device 1 name:** Electric Fan
######Power rating (watts): 75
######Hours used: 8
######Energy consumption: 0.60 kWh
######Classification: Moderate
######**Device 2 name:** Computer
######Power rating (watts): 150
######Hours used: 5
######Energy consumption: 0.75 kWh
######Classification: Moderate
######**Device 3 name:** Projector
######Power rating (watts): 300
######Hours used: 4
######Energy consumption: 1.20 kWh
######Classification: Moderate
######**Total energy consumption:** 2.55 kWh


> **Concepts:** Numeric and string arguments, numeric and string return values, function composition, for loops, accumulators, and maximum-value tracking.

```python
def calculate_energy(power, hours):
    return power * hours / 1000

def classify_usage(energy):
    if energy < 0.5:
        return "Low"
    elif energy <= 2:
        return "Moderate"
    else:
        return "High"

devices = int(input("How many devices? "))
total = 0
highest = 0
highest_name = ""

for number in range(1, devices + 1):
    name = input("Device name: ")
    power = float(input("Power rating (watts): "))
    hours = float(input("Hours used: "))

    energy = calculate_energy(power, hours)
    print("Energy consumption:", energy, "kWh")
    print("Classification:", classify_usage(energy))

    total = total + energy

    if energy > highest:
        highest = energy
        highest_name = name

print("Total energy consumption:", total, "kWh")
print("Device with highest consumption:", highest_name)
```

##**3. Water Refilling Station Sales Tracker**
###**Scenario**
A water refilling station wants to track its sales for the day. The cashier enters the number of containers sold for each transaction until the transaction ends.
###**Requirements**
- Create a function calculate_sale(quantity, price_per_container) that returns the total sale for one transaction.
- Create a function calculate_discount(total) that returns a 5% discount if the transaction is worth at least ₱500; otherwise, it returns 0.
- Ask the user to enter the price per container.
- Use a while loop to accept the quantity sold in each transaction. Stop when the quantity is 0.
- For each transaction, display the original sale, discount, and final amount.
- Accumulate the total sales before discounts and the total discounts.
- Display the final sales summary.
###**Sample Output**
######Price per container: ₱30
######**Enter quantity sold (0 to finish):** 10
######Sale: ₱300.00
######Discount: ₱0.00
######Amount due: ₱300.00
######**Enter quantity sold (0 to finish):** 20
######Sale: ₱600.00
######Discount: ₱30.00
######Amount due: ₱570.00
######**Enter quantity sold (0 to finish):** 0
######Total sales before discounts: ₱900.00
######Total discounts: ₱30.00
######Net sales: ₱870.00
> **Concepts:** Multiple function calls, numeric return values, conditional statements, while loops, sentinel-controlled iteration, and accumulators.

```python
def calculate_sale(quantity, price):
    return quantity * price

def calculate_discount(total):
    if total >= 500:
        return total * 0.05
    else:
        return 0

price = float(input("Price per container: ₱"))
sales = 0
discounts = 0

quantity = float(input("Enter quantity (0 to finish): "))

while quantity != 0:
    sale = calculate_sale(quantity, price)
    discount = calculate_discount(sale)

    print("Sale: ₱", sale)
    print("Discount: ₱", discount)
    print("Amount due: ₱", sale - discount)

    sales = sales + sale
    discounts = discounts + discount

    quantity = float(input("Enter quantity (0 to finish): "))

print("Total sales before discounts: ₱", sales)
print("Total discounts: ₱", discounts)
print("Net sales: ₱", sales - discounts)
```

##**5. School Event Attendance Monitor**
###**Scenario**
An event organizer needs a program to record attendance for several sessions of a school event. The program compares the number of attendees against the expected attendance for each session.
###**Requirements**
- Create a function calculate_attendance_rate(attendees, expected) that returns the attendance percentage. If expected is zero, return 0.
- Create a function classify_attendance(rate) that returns:
- "Low" if the rate is below 60%.
- "Moderate" if the rate is from 60% to below 90%.
- "High" if the rate is 90% or above.
- Ask the user how many event sessions will be monitored.
- Use a for loop to enter the expected attendance and actual number of attendees for each session.
- Display each session's attendance rate and classification.
- Count how many sessions have "High" attendance.
- Calculate and display the total number of attendees and the overall attendance rate based on the combined attendance and expected attendance across all sessions.
###**Sample Output**
######How many sessions? 3
######**Session 1 expected attendance:** 100
######Actual attendees: 95
######Attendance rate: 95.00%
######Classification: High
######**Session 2 expected attendance:** 80
######Actual attendees: 60
######Attendance rate: 75.00%
######Classification: Moderate
######**Session 3 expected attendance:** 120
######Actual attendees: 50
######Attendance rate: 41.67%
######Classification: Low
######**Total attendees:** 205
######**Total expected attendance:** 300
######**Overall attendance rate:** 68.33%
######**High-attendance sessions:** 1

> **Concepts:** Functions with numeric and string return values, conditional statements, for loops, counters, accumulators, and percentage calculations.

```python
def calculate_attendance_rate(attendees, expected):
    if expected == 0:
        return 0
    return attendees / expected * 100

def classify_attendance(rate):
    if rate < 60:
        return "Low"
    elif rate < 90:
        return "Moderate"
    else:
        return "High"

sessions = float(input("How many sessions? "))
total_attendees = 0
total_expected = 0
high = 0
number = 1

while number <= sessions:
    expected = float(input("Expected attendance: "))
    attendees = float(input("Actual attendees: "))

    rate = calculate_attendance_rate(attendees, expected)
    print("Attendance rate:", round(rate, 2))
    print("Classification:", classify_attendance(rate))

    if rate >= 90:
        high = high + 1

    total_attendees = total_attendees + attendees
    total_expected = total_expected + expected
    number = number + 1

print("Total attendees:", total_attendees)
print("Total expected:", total_expected)
print("Overall rate:", round(calculate_attendance_rate(total_attendees, total_expected), 2))
print("High-attendance sessions:", high)
```
